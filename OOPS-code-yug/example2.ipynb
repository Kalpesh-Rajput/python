# Access Members of One Class Inside Another Class

```python
class Employee:
    def __init__(self,eid,name,sal):
        self.emp_id = eid
        self.emp_name = name
        self.emp_salary = sal

    def display(self):
        print("emplyee id :",self.emp_id)
        print("emplyee name :",self.emp_name)
        print("emplyee salary :",self.emp_salary)


class Changes:
    @staticmethod
    def increment(obj):
        obj.emp_salary = obj.emp_salary+2000
        obj.display()


e1 = Employee(101,'shantanu',50000)
Changes.increment(e1)
```

# Destructor


```python
import time 
class Employee:
    def __init__(self,name,sal):
        self.emp_name = name
        self.emp_salary = sal

    def display(self):
        print(f"name is {self.name} and salary is {self.salary}")
    
    def __del__(self):
        print("destructor is called")

    
e1 = Employee("shantanu",50000)
e2 = e1

del e1
time.sleep(5)
```

```python
class Movie(object):
    def __init__(self,title,mins,hero):
        self.title = title
        self.runtime = title
        self.hero = title
    def printer(self):
        print(f"title is {self.title}\nrunetime: {self.runtime}\nhero {self.hero}")
list_of_movies = []
while True:
    title = input("enter the title of movie")
    mins = input("enter the title of movie")
    hero = input("enter the title of movie")
    obj = Movie(title,mins,hero)
    list_of_movies.append(obj)
    print("movie added into the list")
    ans = input("do you want to add another movies") 
    if ans!= "y":
        break

print("all movies ionfo")
for obj in list_of_movies:
    obj.printer()    
```
