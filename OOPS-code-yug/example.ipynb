```python
# Buil-in Class functions
class Employee:
    def __init__(self,sal,ag):
        self.salary = sal
        self.age = ag
        
    def Display(self):
        print(f"the salary of employee is {self.salary} and age is {self.age}")
        

e1 = Employee(25000,25)
e2 = Employee(15000,20)

print(getattr(e1,'salary'))
print(getattr(e1,'age'))
setattr(e2,'salary',50000)
print(e2.__dict__)
delattr(e2,'age')
print(e2.__dict__)
print(hasattr(e1,'salary'))
print(hasattr(e2,'age'))
```

```python
# Inheritance

class Employee:
    bonus=5000
    def display(self):
        print("this is employee class method")
    
class Manager(Employee):# inheritance
    bonus1=10000
    def show(self):
        print("this is manager class method")
        

e1 = Employee()
m1 = Manager()

e1.display()
m1.show()

print(m1.bonus) # manager can access employee class attributes
print(m1.bonus1)
print(e1.bonus)
# print(e1.bonus1) # employee can not access manager class attribute
```

```python
# Multilevel Inheritance
class Human_being:
    def __init__(self):
        print("Human being constructor called")
        self.name = "kalpesh"

class Employee(Human_being):
    def __init__(self):
        print("Employee constructor called")
        self.bonus = 25000
        super().__init__() # to call parent class constructor
    
class manager(Employee):
    def __init__(self):
        print("Manager constructor called")
        self.bonus1 = 50000
        super().__init__() # to call parent class constructor


e1 = manager()
print(e1.__dict__)
```

```python
#Multiple Inheritance
class Country:
    def __init__(self):
        print("country constructor called")
        self.office = "Delhi"

class State:
    def __init__(self):
        super().__init__() # to call parent class constructor
        print("state constructor called")
        self.office = "Mumbai"

class District(State, Country):
    def __init__(self):
        super().__init__() # to call parent class constructor
        print("city constructor called")
        self.office = "Pune"

    
d= District()
print(d.office) # it will print pune because of MRO (Method Resolution Order) in multiple inheritance
    
```

```python
# Encapsulation

class finance:
    def __init__(self):
        self.__revenue = 25000 # private variable
        self._sales = 14 # protected variable

f1 = finance()
print(f1.__dict__) # we can not access private variable outside the class 
class HR:
    def __init__(self):
        self.number_of_employees = 100
        self.__revenue = 50 

h1 = HR()
print(f1.__dict__)
```
